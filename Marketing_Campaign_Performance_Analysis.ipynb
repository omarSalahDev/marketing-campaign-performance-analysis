```python
clustering_df = segmentation_df[
    segmentation_df["Income"] < 666666
].copy()

clustering_df.shape
```

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

scaled_data = scaler.fit_transform(clustering_df)

scaled_data[:5]
```

```python
from sklearn.cluster import KMeans
import matplotlib.pyplot as plt

inertia = []

for k in range(2, 9):
    kmeans = KMeans(
        n_clusters=k,
        random_state=42,
        n_init=10
    )
    kmeans.fit(scaled_data)
    inertia.append(kmeans.inertia_)

plt.figure(figsize=(8, 5))
plt.plot(range(2, 9), inertia, marker="o")
plt.title("Elbow Method for Optimal Number of Clusters")
plt.xlabel("Number of Clusters (K)")
plt.ylabel("Inertia")
plt.xticks(range(2, 9))
plt.show()
```

```python
from sklearn.metrics import silhouette_score

silhouette_scores = []

for k in range(2, 9):
    kmeans = KMeans(
        n_clusters=k,
        random_state=42,
        n_init=10
    )

    labels = kmeans.fit_predict(scaled_data)
    score = silhouette_score(scaled_data, labels)

    silhouette_scores.append(score)

for k, score in zip(range(2, 9), silhouette_scores):
    print(f"K={k}: Silhouette Score = {score:.3f}")
```

```python
kmeans = KMeans(
    n_clusters=2,
    random_state=42,
    n_init=10
)

clustering_df["Cluster"] = kmeans.fit_predict(scaled_data)

clustering_df["Cluster"].value_counts().sort_index()
```

```python
cluster_profile = (
    clustering_df
    .groupby("Cluster")
    [
        [
            "Income",
            "Recency",
            "Total_Spending",
            "Total_Purchases",
            "NumWebPurchases",
            "NumCatalogPurchases",
            "NumStorePurchases"
        ]
    ]
    .mean()
    .round(2)
)

cluster_profile
```

```python
cluster_analysis = clustering_df[["Cluster"]].copy()

cluster_analysis = cluster_analysis.join(
    analysis_df["Response"]
)

cluster_response = (
    cluster_analysis
    .groupby("Cluster")["Response"]
    .agg(
        total_customers="count",
        responding_customers="sum",
        response_rate="mean"
    )
)

cluster_response["response_rate"] = (
    cluster_response["response_rate"] * 100
).round(2)

cluster_response
```

```python
cluster_profile_full = (
    clustering_df
    .join(
        analysis_df[
            [
                "Age",
                "Education",
                "Children",
                "Family_Size",
                "Customer_Tenure_Days",
                "Response",
                "AcceptedCmp1",
                "AcceptedCmp2",
                "AcceptedCmp3",
                "AcceptedCmp4",
                "AcceptedCmp5"
            ]
        ]
    )
    .groupby("Cluster")
    .agg({
        "Age": "mean",
        "Children": "mean",
        "Family_Size": "mean",
        "Customer_Tenure_Days": "mean",
        "Response": "mean",
        "AcceptedCmp1": "mean",
        "AcceptedCmp2": "mean",
        "AcceptedCmp3": "mean",
        "AcceptedCmp4": "mean",
        "AcceptedCmp5": "mean"
    })
    .round(2)
)

cluster_profile_full
```

```python
final_segment_profile = (
    clustering_df
    .groupby("Cluster")
    .agg(
        Customers=("Income", "count"),
        Avg_Income=("Income", "mean"),
        Avg_Recency=("Recency", "mean"),
        Avg_Spending=("Total_Spending", "mean"),
        Avg_Purchases=("Total_Purchases", "mean"),
        Avg_Web_Purchases=("NumWebPurchases", "mean"),
        Avg_Catalog_Purchases=("NumCatalogPurchases", "mean"),
        Avg_Store_Purchases=("NumStorePurchases", "mean")
    )
)

# Add response rate
final_segment_profile["Response_Rate"] = (
    cluster_response["response_rate"]
)

# Add customer percentage
final_segment_profile["Customer_Share"] = (
    final_segment_profile["Customers"]
    / final_segment_profile["Customers"].sum()
    * 100
)

final_segment_profile.round(2)
```

```python
# Insight 1: Customer Value vs Campaign Response

cluster_response_summary = final_segment_profile[
    [
        "Customers",
        "Avg_Income",
        "Avg_Spending",
        "Avg_Purchases",
        "Response_Rate"
    ]
].copy()

cluster_response_summary
```

```python
lower_response = final_segment_profile.loc[0, "Response_Rate"]
higher_response = final_segment_profile.loc[1, "Response_Rate"]

response_difference = higher_response - lower_response
response_lift = higher_response / lower_response

print(f"Response rate difference: {response_difference:.2f} percentage points")
print(f"Response lift: {response_lift:.2f}x")
```

```python
# Validate customer segments

segment_check = (
    final_analysis_df
    .groupby("Customer_Segment")
    .agg(
        Customers=("ID", "count"),
        Responders=("Response", "sum"),
        Avg_Spending=("Total_Spending", "mean"),
        Avg_Purchases=("Total_Purchases", "mean"),
        Avg_Income=("Income", "mean")
    )
)

segment_check["Response_Rate"] = (
    segment_check["Responders"]
    / segment_check["Customers"]
    * 100
).round(2)

segment_check.round(2)
```
